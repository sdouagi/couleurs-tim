## Principe

L’algorithme **SLIC** est utilisé pour générer des **superpixels** dans une image. Les superpixels regroupent des pixels voisins ayant des caractéristiques similaires, ce qui permet de réduire la complexité des traitements d’image ultérieurs (segmentation, reconnaissance d’objets, etc.).

---

## 1. Objectif

SLIC regroupe les pixels en **clusters homogènes** en couleur et proximité spatiale, tout en respectant les contours des objets dans l’image. Chaque cluster correspond à un superpixel.

---

## 2. Représentation des pixels

Chaque pixel est représenté dans un espace **5D** :  

$
[x, y, L, a, b]
$

- $x, y$ : coordonnées spatiales du pixel  
- $L, a, b$ : valeurs de couleur dans l’espace CIELAB (perception de couleur plus fidèle que RGB)

---

## 3. Initialisation

1. Choisir le **nombre de superpixels** $K$  
2. Calculer la **taille approximative de chaque superpixel** $S$ :  

$
S = \sqrt{\frac{N}{K}}
$  

où $N$ est le nombre total de pixels dans l’image.  

3. Placer les **centroïdes initiaux** sur une grille régulière espacée de $S$ pixels.  
4. Ajuster la position des centroïdes vers un minimum de gradient local pour éviter qu’ils ne tombent sur des bords forts.

---

## 4. Assignment des pixels

Pour chaque centroïde $C_k$ :  

1. Considérer seulement les pixels dans une **fenêtre $2S \times 2S$** autour du centroïde.  
2. Calculer une **distance combinée** $D$ pour chaque pixel $i$ :

$
D = \sqrt{ \left(\frac{d_c}{m}\right)^2 + \left(\frac{d_s}{S}\right)^2 }
$

où :  
- $d_c = \sqrt{(L_i - L_k)^2 + (a_i - a_k)^2 + (b_i - b_k)^2}$ : distance couleur  
- $d_s = \sqrt{(x_i - x_k)^2 + (y_i - y_k)^2}$ : distance spatiale  
- $m$ : paramètre de **compacité** (équilibre entre couleur et proximité spatiale)  

3. Assigner chaque pixel au centroïde pour lequel $D$ est minimal.

---

## 5. Mise à jour des centroïdes

- Chaque centroïde devient la **moyenne** des pixels qui lui sont assignés (coordonnées spatiales et couleur).

---

## 6. Itération

- Répéter les étapes **assignment** et **mise à jour** jusqu’à convergence (souvent ~10 itérations suffisent).

---

## 7. Post-traitement

- Pour garantir que les superpixels soient **connexes**, on peut réassigner les pixels isolés aux superpixels voisins.

---

## Points clés

- **Rapide** : la recherche des pixels est limitée à un voisinage $2S \times 2S$  
- **Superpixels réguliers** mais flexibles pour suivre les contours  
- **Paramètre $m$** : contrôle la régularité vs fidélité aux contours

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Charger l'image
path_images = "../Images_Projet/"
img = cv2.imread(path_images+'fleur.jpg', 1)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convertir en float et en espace Lab (OpenCV)
img_lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB).astype(np.float64)

plt.imshow(img_rgb);
```

## Centroïde 

```python
# Paramètres
K = 200
H, W = img_lab.shape[:2]
S = int(np.sqrt(H*W / K))

centroids = []
for i in range(S//2, H, S):
    for j in range(S//2, W, S):
        L, a, b = img_lab[i, j]
        centroids.append([L, a, b, i, j])
centroids = np.array(centroids)
```

## Fonction distance combinée (voir la formule du foutu PDF)

```python
def compute_distance(pixel_lab, pixel_xy, centroid, S, m=10):
    d_c = np.linalg.norm(pixel_lab - centroid[:3])
    d_s = np.linalg.norm(pixel_xy - centroid[3:5])
    D = np.sqrt((d_c/m)**2 + (d_s/S)**2)
    return D
```

## Assignation et mise à jour

```python
def update_centroids(img_lab, labels, n_centroids):
    new_centroids = []
    for k in range(n_centroids):
        indices = np.argwhere(labels == k)
        if len(indices) == 0:
            continue
        L_mean = np.mean(img_lab[indices[:,0], indices[:,1], 0])
        a_mean = np.mean(img_lab[indices[:,0], indices[:,1], 1])
        b_mean = np.mean(img_lab[indices[:,0], indices[:,1], 2])
        x_mean = np.mean(indices[:,0])
        y_mean = np.mean(indices[:,1])
        new_centroids.append([L_mean, a_mean, b_mean, x_mean, y_mean])
    return np.array(new_centroids)
```

## Boucle itérative

```python
num_iter = 10
m = 10
labels = -1 * np.ones((H, W), dtype=int)
distances = np.full((H, W), np.inf)

for it in range(num_iter):
    print(f"Iteration {it+1}/{num_iter}")
    
    distances.fill(np.inf)
    labels.fill(-1)
    
    for k, c in enumerate(centroids):
        x_c, y_c = int(c[3]), int(c[4])
        x_min = max(x_c - S, 0)
        x_max = min(x_c + S, H)
        y_min = max(y_c - S, 0)
        y_max = min(y_c + S, W)
        
        for i in range(x_min, x_max):
            for j in range(y_min, y_max):
                pixel_lab = img_lab[i, j]
                pixel_xy = np.array([i, j])
                D = compute_distance(pixel_lab, pixel_xy, c, S, m)
                
                if D < distances[i, j]:
                    distances[i, j] = D
                    labels[i, j] = k
    
    centroids = update_centroids(img_lab, labels, len(centroids))
```

## Tracer les contours des superpixels

```python
contour_img = img_rgb.copy()
for i in range(1, H-1):
    for j in range(1, W-1):
        if (labels[i,j] != labels[i+1,j] or 
            labels[i,j] != labels[i-1,j] or
            labels[i,j] != labels[i,j+1] or
            labels[i,j] != labels[i,j-1]):
            contour_img[i,j] = [255, 0, 0]  # rouge

plt.figure(figsize=(8,8))
plt.imshow(contour_img)
plt.axis('off')
plt.title("Superpixels SLIC faits à la main (sans scikit-image)")
```
